```python
import xml.etree.ElementTree as et
tree = et.ElementTree(file='menu.xml')  # 올린 파일 이름
root = tree.getroot()

print(root.tag)
```

```python
for Child in root:
  print('tag:', Child.tag, 'attributes:',Child.attrib)
  for grandchild in Child:
    print('\ttag:', grandchild.tag, 'attributes"', grandchild.attrib)
```

```python
print(root[0])                    # root의 첫 번째 요소 출력
print(root[0].get("hours"))       # "hours" 키의 값 출력
print(root[0].keys())             # 딕셔너리의 모든 키를 리스트로 리턴
print(root[0].items())            # 모든 속성의 key, Value 쌍을 리턴
```

```python
lunch = root.find("lunch")
print(lunch)
print(lunch.get("hours"))
print(lunch.keys())
print(lunch.items())
```

```python
print(root[0].findall('item'))
```

```python
print(root[0].findall('item'))

for item in root[0].findall('item'):
    print(item.get('price'))
    print(item.text)
```
