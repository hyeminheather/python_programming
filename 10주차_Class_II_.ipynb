class 실질예

```python
# 계산기 프로그램
 # 계산기 더하기 기능
 #함수로 짜는 경우

result1=0
result2=0

def adder1(num):
  global result1
  result1 = result1 + num
  return result1

def adder2(num):
  global result2
  result2 = result2 + num
  return result2
```

```python
adder1(6)
```

```python
adder1(3)
```

```python
adder2(3)
```

```python
adder2(4)
```

```python
from typing import Callable
#여러개 계산기가 필요하면 클래스를 사용하면 됨

class Calculator:
  def __init__(self):
    self.result = 0

  def adder(self, num):
    self.result += num
    return self.result

```

```python
cal1 = Calculator()  #객체를 만듬
cal2 = Calculator()
```

```python
cal1.adder(3) # 클래스 내부에 있는 함수는, 함수라고 해도되지만, 메서드라고 부른다
```

```python
cal1.adder(5)
```

```python
cal2.adder(5)
```

```python
cal2.adder(9)
```

```python
class FourCal:
  def setdata(self, first, second):
    self.first = first
    self.second = second

  def sum(self):
    result = self.first + self.second
    return result

  def mul(self):
    result = self.first * self.second
    return result

  def sub(self):
    result = self.first - self.second
    return result


  def divide(self):
    result = self.first - self.second
    return result
```

```python
a = FourCal()
```

```python
a.setdata(4,2)
```
